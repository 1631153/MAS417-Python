```python
import mujoco
import numpy as np
import mediapy as media
print(f"MuJoCo {mujoco.__version__} ready!")

# A pendulum: one body, connected to the world by a hinge joint
pendulum_xml = """
<mujoco>
  <option gravity="0 0 -9.81"/>
  <worldbody>
    <light pos="0 -1 2" dir="0 0.5 -1"/>
    
    <!-- The pendulum arm -->
    <body name="arm" pos="0 0 1.5">
      <!-- THIS is what makes it swing: a hinge joint -->
      <!-- Without this joint, the arm would be frozen in place -->
      <joint name="hinge" type="hinge" axis="0 1 0"/>
      
      <!-- The visible/physical shape -->
      <geom type="box" fromto="0 0 0 0 0 -0.5" size="0.04" 
            mass="1" rgba="0.2 0.6 0.9 1"/>
      <geom type="box" fromto="0 0 0 0 0 0.2" size="0.06" 
            mass="1" rgba="0.2 0.6 0.9 1"/>
      <!-- The pendulum bob -->
      <geom type="box" fromto="0 0 -0.5 0 0 -0.7" size="0.08" 
            mass="2" rgba="0.5 0.6 0.9 1"/>
    </body name="ground">
       <geom type="plane" pos="0 0 0" size="2 2 0.1" rgba="0.9 0.9 0.9 1"/>
       
    <body name="stand" pos="0 0.2 1.5">
       <geom type="box" fromto="0 0 0 0 0 -1.5" size="0.04" 
            mass="1" rgba="0.9 0.6 0.9 1"/>
       <geom type="box" fromto="0 -0.4 0 0 -0.4 -1.5" size="0.04" 
            mass="1" rgba="0.9 0.6 0.9 1"/>
    </body>
  </worldbody>
</mujoco>
"""

m = mujoco.MjModel.from_xml_string(pendulum_xml)
d = mujoco.MjData(m)

# Start the pendulum at 90 degrees
d.qpos[0] = np.pi / 4 # 90° in radians

# Simulate
renderer = mujoco.Renderer(m, height=400, width=400)
frames = []
while d.time < 7.0:
    mujoco.mj_step(m, d)
    if len(frames) < d.time * 30:
        renderer.update_scene(d)
        frames.append(renderer.render().copy())
renderer.close()

media.show_video(frames, fps=30)
```
